```python
from dotenv import load_dotenv

load_dotenv()
import os

hub_name = os.environ["HUB_NAME"]
```

```python
from huggingface_hub import login
from datasets import load_dataset

login()

dataset = load_dataset("nax0/LaLoMa-dataset", streaming=False)

dataset_train = dataset["train"].shuffle()
dataset_test = dataset["test"]
samples_eval = dataset["samples_eval"]
samples_test = dataset["sample_test"]
```

```python
from tqdm.auto import tqdm

information = []
for sample in tqdm(dataset_train):
    if str(sample["order"]) != "1":
        continue
    if (len(sample["text"]) < 150) or (len(sample["text"]) > 200):
        continue
    information.append(sample)

information = sorted(information, key=lambda x: x["page"])
```

```python
print(len(information))
```

```python
from IPython.display import display, HTML


def _print_latex(latex_text):
    html_string = f"""
    <div style='font-size: 50px; color: white; background-color: #1e1e1e; padding: 15px; border-radius: 8px;'>
        {latex_text}
    </div>
    """
    display(HTML(html_string))


for sample in information:
    print("\n\n\n" + "-" * 120)
    print(
        f"(L{len(sample["text"])}) Sample of page {sample["page"]}, order {sample["order"]}, ann_id = {sample["ann_id"]}"
    )
    _print_latex(sample["text"])
    print(sample["text"])
    display(sample["image"])
    print("\n\n\n")
```
